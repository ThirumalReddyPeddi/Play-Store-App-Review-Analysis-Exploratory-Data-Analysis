##### Q).How do average ratings differ between free and paid apps?



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Rating to numeric
playstore_data['Rating'] = pd.to_numeric(playstore_data['Rating'], errors='coerce')

# Keep valid ratings and app types
rating_df = playstore_data[
    playstore_data['Rating'].between(1, 5) &
    playstore_data['Type'].isin(['Free', 'Paid'])
].copy()

# Calculate average rating
avg_rating = (
    rating_df.groupby('Type')['Rating']
    .mean()
    .reset_index()
)

# Plot grouped bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    data=avg_rating,
    x='Type',
    y='Rating',
    hue='Type',
    palette='Set2',
    legend=False
)

plt.title('Average Ratings: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.tight_layout()
plt.show()

print(avg_rating)
```

###### 1. Why did you pick the specific chart?

A grouped bar chart was selected because it makes it easy to compare the average ratings of free and paid applications side by side. It clearly shows whether one app type has a higher average user rating than the other.
