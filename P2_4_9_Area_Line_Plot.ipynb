#####Q).How has the number of app updates changed over time?

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert Last Updated to datetime
playstore_data['Last Updated'] = pd.to_datetime(
    playstore_data['Last Updated'],
    errors='coerce'
)

# Remove missing dates
update_df = playstore_data.dropna(subset=['Last Updated']).copy()

# Extract year
update_df['Update_Year'] = update_df['Last Updated'].dt.year

# Count apps by last update year
yearly_updates = (
    update_df.groupby('Update_Year')['App']
    .nunique()
    .sort_index()
)

# Area line plot
plt.figure(figsize=(12, 6))

plt.fill_between(
    yearly_updates.index,
    yearly_updates.values,
    alpha=0.3
)

plt.plot(
    yearly_updates.index,
    yearly_updates.values,
    marker='o',
    linewidth=2
)

plt.title('Number of Apps by Year of Last Update')
plt.xlabel('Year of Last Update')
plt.ylabel('Number of Apps')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(yearly_updates)
```

###### 1. Why did you pick the specific chart?

An area line plot was selected because it helps visualize changes in the number of applications across different years. The shaded area emphasizes the magnitude of applications associated with each last-update year and makes upward or downward trends easier to identify.

Identify the years in which the highest and lowest numbers of apps were last updated.

Understand how the number of apps varies across different update years.

Observe whether the number of applications with recent updates is increasing or decreasing.
