### Playstore Data Variables Description



| Variable | Description |
|---|---|
| **App** | Name of the application. |
| **Category** | Category to which the application belongs. |
| **Rating** | Average user rating of the application. |
| **Reviews** | Number of user reviews received by the application. |
| **Size** | Size of the application. |
| **Installs** | Number of times the application has been installed. |
| **Type** | Indicates whether the application is Free or Paid. |
| **Price** | Price of the application. |
| **Content Rating** | Target age/content group of the application, such as Everyone, Teen, or Mature 17+. |
| **Genres** | Genre or sub-category of the application. |
| **Last Updated** | Date when the application was last updated. |
| **Current Ver** | Current version of the application. |
| **Android Ver** | Minimum Android version required to run the application. |

### User reviews Variables Description



| Variable | Description |
|---|---|
| **App** | Name of the application for which the user review was given. |
| **Translated_Review** | User review text translated into English. |
| **Sentiment** | Overall sentiment of the user review, such as Positive, Negative, or Neutral. |
| **Sentiment_Polarity** | Measures the direction and strength of the sentiment, ranging from negative to positive. |
| **Sentiment_Subjectivity** | Measures how subjective or opinion-based the review is. |
