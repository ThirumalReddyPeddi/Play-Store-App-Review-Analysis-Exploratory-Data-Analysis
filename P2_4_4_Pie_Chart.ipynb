#####Q).What percentage of apps belong to each app type (Free vs. Paid)?

```python
playstore_data['Type'].unique()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Count Free and Paid apps
app_type_count = playstore_data['Type'].dropna().str.strip().str.title().value_counts()

# Calculate percentages
app_type_percentage = app_type_count / app_type_count.sum() * 100

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    app_type_percentage,
    labels=app_type_percentage.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0.05] * len(app_type_percentage)
)

plt.title('Percentage Distribution of Free and Paid Apps')
plt.axis('equal')
plt.show()

# Display percentages
print(app_type_percentage.round(2))
```

###### 1. Why did you pick the specific chart?



A pie chart was selected because it effectively represents the percentage distribution of free and paid applications. It makes it easy to compare the proportion of each app type and understand which type dominates the Play Store.

###### 2. What is/are the insight(s) found from the chart?

The pie chart shows the proportion of free and paid applications in the dataset. It helps identify whether free or paid applications make up the larger share of the Play Store. This provides an understanding of the dominant pricing model.
