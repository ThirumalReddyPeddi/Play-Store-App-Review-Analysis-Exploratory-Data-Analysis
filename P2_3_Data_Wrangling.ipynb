**1.Ratings Column**

```python
import pandas as pd
#Convert Rating to numeric
playstore_data['Rating'] = pd.to_numeric(playstore_data['Rating'], errors='coerce')
playstore_data.head()
```

**2.Last Updated Column**

```python
# Changing the datatype of the Last Updated column from string to datetime.
# When the values in the last updated column are passed via Pandas' to datetime() function, the result is a string Date time that is converted into a Python Date time object.
playstore_data["Last Updated"] = pd.to_datetime(playstore_data['Last Updated'])
playstore_data.head()
```

**3.Reviews Column:-**

```python
# Changing the Reviews column's values' datatype from string to int.
playstore_data['Reviews'] = playstore_data['Reviews'].astype(int)
playstore_data.head()
```

```python
playstore_data.dtypes
```

**4.Installs column**

```python

# Changing the Installs column's values from string to integer datatype.
# Checking the contents of the 'Installs' column
playstore_data['Installs'].value_counts()
```

```python
playstore_data.dtypes
```

```python
# 4. Clean and convert Installs to numeric
playstore_data['Installs'] = (
    playstore_data['Installs']
    .astype(str)
    .str.replace('+', '', regex=False)
    .str.replace(',', '', regex=False)
)

playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'], errors='coerce'
)
```

```python
playstore_data.dtypes
```

**5.Price Column:-**

```python
# 5. Clean Price and convert it to numeric
playstore_data['Price'] = (
    playstore_data['Price']
    .astype(str)
    .str.replace('$', '', regex=False)
)

playstore_data['Price'] = pd.to_numeric(
    playstore_data['Price'], errors='coerce'
)
```
