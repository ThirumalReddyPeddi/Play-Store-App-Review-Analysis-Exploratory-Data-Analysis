#####Q).What is the distribution of app ratings?

```python
# Convert Rating into numeric format
rating_data = pd.to_numeric(playstore_data['Rating'], errors='coerce')

# Remove missing and invalid ratings
rating_data = rating_data[rating_data.between(1, 5)]

# Create histogram
plt.figure(figsize=(9, 5))

sns.histplot(
    rating_data,
    bins=20,
    kde=True
)

plt.title('Distribution of App Ratings')
plt.xlabel('App Rating')
plt.ylabel('Number of Applications')
plt.xlim(1, 5)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# Display summary statistics
print(rating_data.describe().round(2))
```

```python
# What this chart shows:
    # - Most frequently occurring app ratings.
    # - Whether ratings are concentrated around higher or lower values.
    # - Whether the distribution is symmetric or skewed.
    # - The overall spread of app ratings.
```

###### 1. Why did you pick the specific chart?

A histogram was selected because it helps visualize the frequency distribution of app ratings. It makes it easy to understand how ratings are spread across different ranges and identify the most frequently occurring ratings. It also helps identify whether ratings are concentrated toward higher or lower values.

###### 2. What is/are the insight(s) found from the chart?




**The histogram helps identify:**

Whether most applications have high, moderate, or low ratings.

Whether the rating distribution is left-skewed, right-skewed, or approximately symmetric.

The overall variation in app ratings.

These observations provide an understanding of overall app quality and user satisfaction.
