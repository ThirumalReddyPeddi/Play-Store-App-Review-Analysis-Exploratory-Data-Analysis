#### Chart - 1:**Line Plot**

##### Q).How has the average app rating changed over time based on the last updated date?

````python
# Chart - 1 visualization code

# Average App Rating by Last Updated Year
````

````python
import pandas as pd
import matplotlib.pyplot as plt

# Convert Last Updated column to datetime
playstore_data['Last Updated'] = pd.to_datetime(
    playstore_data['Last Updated'],
    errors='coerce'
)

# Keep valid ratings and update dates
rating_df = playstore_data[
    playstore_data['Rating'].between(1, 5) &
    playstore_data['Last Updated'].notna()
].copy()

# Extract year from the last update date
rating_df['Update_Year'] = (
    rating_df['Last Updated'].dt.year
)

# Calculate average rating for each year
yearly_rating = (
    rating_df.groupby('Update_Year')['Rating']
    .mean()
    .sort_index()
)

# Plot the line chart
plt.figure(figsize=(10, 5))

plt.plot(
    yearly_rating.index,
    yearly_rating.values,
    marker='o',
    linewidth=2
)

plt.title('Average App Rating by Last Updated Year')
plt.xlabel('Year of Last Update')
plt.ylabel('Average Rating')
plt.ylim(1, 5)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

###### 1. Why did you pick the specific chart?

A line chart was selected because it helps visualize changes in average app ratings across different years. It makes it easier to identify upward trends, downward trends, and periods of stable ratings based on the year in which apps were last updated.

###### 2. What is/are the insight(s) found from the chart?

The chart shows the average ratings of apps grouped by their last-update year. An upward trend indicates higher average ratings among apps last updated in those years, while a downward trend indicates lower average ratings. The actual trend and the years showing the highest and lowest average ratings should be identified from the generated chart.

##### Q).What is the relationship between app ratings and the number of reviews?

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select required columns
data = playstore_data[['Rating', 'Reviews']].copy()

# Convert columns into numeric format
data['Rating'] = pd.to_numeric(
    data['Rating'], errors='coerce'
)

data['Reviews'] = pd.to_numeric(
    data['Reviews'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

# Remove missing and invalid values
data = data.dropna()

data = data[
    data['Rating'].between(1, 5) &
    (data['Reviews'] > 0)
]

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=data,
    x='Reviews',
    y='Rating',
    alpha=0.5
)

plt.xscale('log')
plt.title('Relationship Between App Ratings and Number of Reviews')
plt.xlabel('Number of Reviews (Log Scale)')
plt.ylabel('App Rating')
plt.ylim(1, 5)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Calculate correlation
correlation = data['Rating'].corr(data['Reviews'])

print("Correlation:", round(correlation, 3))
````

###### 1. Why did you pick the specific chart?

A scatter plot was selected because it helps visualize the relationship between app ratings and the number of reviews. Each point represents an individual app, making it easier to identify patterns, clusters, and outliers. It also helps determine whether apps with a higher number of reviews tend to receive higher or lower ratings.

###### 2. What is/are the insight(s) found from the chart?

The scatter plot illustrates the relationship between app ratings and the number of reviews. The correlation coefficient helps determine the direction and strength of this relationship. The chart also helps identify apps with a high number of reviews but relatively low ratings, which may indicate potential user dissatisfaction.

Overall, the relationship should be interpreted based on the observed scatter pattern and calculated correlation value. A higher number of reviews does not necessarily indicate a higher app rating.


###### 3. Will the gained insights help creating a positive business impact?

**Positive Business Impact:**

**Negative Growth Implications:**

#### Chart - 5:**boxplot**

#####Q).How does app size vary across different app categories?

````python
# Chart - 5 visualization code

# Useful for understanding distributions, medians, spread, and outliers in numerical data across different categories.
````

````python
# Function to convert app size into MB
def convert_size(size):
    size = str(size).strip().upper()

    if size.endswith('M'):
        return float(size[:-1])

    elif size.endswith('K'):# The Size column contains values such as 19M, 14M, 512k, and Varies with device.
        return float(size[:-1]) / 1024 # We need to convert these values into numeric MB before plotting.

    else:
        return None

# Create a copy of required columns
size_df = playstore_data[['Category', 'Size']].copy()

# Convert Size into MB
size_df['Size_MB'] = size_df['Size'].apply(convert_size)

# Remove missing values
size_df = size_df.dropna(subset=['Category', 'Size_MB'])

# Create box plotting
plt.figure(figsize=(15, 7))

sns.boxplot(
    data=size_df,
    x='Category',
    y='Size_MB'
)

plt.title('Distribution of App Sizes Across Categories')
plt.xlabel('App Category')
plt.ylabel('App Size (MB)')
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
````

````python
# What this chart shows:Median app size in each category.

# Categories with larger or smaller applications.

# Variation in app sizes within each category.

# Outliers representing unusually large or small applications.
````

###### 1. Why did you pick the specific chart?

A box plot was selected because it helps compare app sizes across different categories. It displays the median size, variation, and outliers, making it easier to identify categories containing larger or smaller applications and understand differences in storage requirements.

###### 2. What is/are the insight(s) found from the chart?

**The box plot helps identify:**

1.Categories with larger and smaller median app sizes.

2.Categories with greater variation in application sizes.

3.Applications with unusually large or small sizes.

These observations help understand storage requirements across different app categories.
