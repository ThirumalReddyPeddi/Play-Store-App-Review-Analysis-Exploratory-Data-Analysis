#### Chart - 7 : **Stacked Bar Chart**

#####Q).How many free and paid apps are present in each app category?

```python
# Chart - 7 visualization code
```

```python
# Select required columns
type_df = playstore_data[['Category', 'Type']].copy()

# Clean app type
type_df['Type'] = (
    type_df['Type'].astype(str).str.strip().str.title()
)

# Remove missing and invalid records
type_df = type_df[
    type_df['Category'].notna() &
    type_df['Type'].isin(['Free', 'Paid'])
]

# Count Free and Paid apps in each category
category_type = pd.crosstab(
    type_df['Category'],
    type_df['Type']
)

# Ensure both columns are present
category_type = category_type.reindex(
    columns=['Free', 'Paid'],
    fill_value=0
)

# Create stacked bar chart
category_type.plot(
    kind='bar',
    stacked=True,
    figsize=(15, 7)
)

plt.title('Distribution of Free and Paid Apps Across Categories')
plt.xlabel('App Category')
plt.ylabel('Number of Applications')
plt.xticks(rotation=90)
plt.legend(title='App Type')
plt.tight_layout()
plt.show()

# Display results
print(category_type)
```

```python
# What this chart shows:
    #- Number of positive, negative, and neutral reviews in each category.
    #- Categories receiving more positive feedback.
    #- Categories receiving more negative feedback.
    #- Differences in user sentiment across app categories.
```
