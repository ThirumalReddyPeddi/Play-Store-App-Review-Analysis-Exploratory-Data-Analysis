##### Q).Which app categories have the highest number of applications?

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count applications in each category
category_counts = playstore_data['Category'].value_counts()

# Plot the top 10 categories
plt.figure(figsize=(18, 9))
sns.barplot(
    x=category_counts.head(10).index,
    y=category_counts.head(10).values
)

plt.title('Top 10 Categories with the Highest Number of Apps')
plt.xlabel('App Category')
plt.ylabel('Number of Applications')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
plt.figure(figsize=(18, 9))

# To display all categories
category_counts.sort_values(ascending=False).plot(kind='bar')


plt.title('Number of Applications by Category')
plt.xlabel('App Category')
plt.ylabel('Number of Applications')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

###### 1. Why did you pick the specific chart?

A **bar chart** was selected because it is effective for comparing the number of applications across different categories. It makes it easy to identify which categories contain the highest and lowest numbers of apps. The chart also provides a clear visual comparison of app distribution across categories.

###### 2. What is/are the insight(s) found from the chart?

The chart shows that the number of applications varies significantly across different categories. Some categories contain a much larger number of apps, while others have relatively fewer applications. The categories with the tallest bars represent the most common app categories in the Google Play Store. These findings help us understand the distribution of applications across different categories.


###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.


Yes, these insights can create a positive business impact by helping developers understand the distribution of apps across different categories. Categories with a high number of applications may indicate strong competition, so developers should focus on innovation, app quality, and user experience to stand out from competitors. Categories with fewer applications may offer potential opportunities for new apps, provided there is sufficient user demand.

However, a high number of applications in a category does not always guarantee high profitability or user satisfaction. Similarly, fewer applications do not necessarily indicate poor growth; they may reflect lower demand. Therefore, developers should consider market demand, ratings, reviews, and competition before making business decisions.

#### Chart - 3

##### Q).Which apps and categories have the highest number of installs?

```python
# Chart - 3 visualization code
```

```python
# 1. Top 10 apps by number of installs

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Convert Installs into numeric format
playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str).str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Top 10 apps by installs
top_apps = (
    playstore_data.groupby('App')['Installs']
    .max()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(14, 7))
sns.barplot(x=top_apps.values, y=top_apps.index)

plt.title('Top 10 Apps by Number of Installs')
plt.xlabel('Number of Installs')
plt.ylabel('App Name')
plt.tight_layout()
plt.show()
```

```python
# 2. Top 10 categories by total installs

category_installs = (
    playstore_data.groupby('Category')['Installs']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(14, 7))
sns.barplot(x=category_installs.values, y=category_installs.index)

plt.title('Top 10 Categories by Total Installs')
plt.xlabel('Total Installs')
plt.ylabel('App Category')
plt.tight_layout()
plt.show()
```

###### 1. Why did you pick the specific chart?

A bar chart was selected because it allows us to compare the number of installs across individual apps and app categories. It clearly highlights the most-installed apps and the categories that account for the highest total installs.

###### 2. What is/are the insight(s) found from the chart?

The charts show differences in install counts among apps and categories. Apps with taller bars have higher install counts, while categories with higher total installs may attract more users overall. These insights help identify popular apps and categories in the Google Play Store.

The exact top apps and categories should be identified from the generated charts or calculated results.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can help businesses identify popular apps and categories, understand market demand, and make better decisions about app development and marketing. Developers can study successful apps to identify useful features and improve their own products.

However, categories with high total installs may also have intense competition, making it difficult for new apps to attract users. Apps with low installs may struggle to gain visibility and generate revenue. Businesses should consider user ratings, reviews, competition, and market demand alongside install counts before deciding where to invest.

High installs do not necessarily guarantee high revenue or user satisfaction.

#####Q).Which content-rating groups have the highest number of applications and installs?

```python
# Chart - 10 visualization code
```

```python
playstore_data['Content Rating'].unique()
```

```python
#

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert installs to numeric
playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Remove rows with missing content rating
content_df = playstore_data.dropna(subset=['Content Rating']).copy()

# Count apps in each content-rating group
content_counts = (
    content_df.groupby('Content Rating')
    .size()
    .sort_values(ascending=False)
)

# Plot number of applications
plt.figure(figsize=(12, 6))
sns.barplot(x=content_counts.index, y=content_counts.values)

plt.title('Number of Applications by Content Rating')
plt.xlabel('Content Rating')
plt.ylabel('Number of Applications')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# 2. Compare total installs by content rating

content_installs = (
    content_df.groupby('Content Rating')['Installs']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))
sns.barplot(x=content_installs.index, y=content_installs.values)

plt.title('Total Installs by Content Rating')
plt.xlabel('Content Rating')
plt.ylabel('Total Installs')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Identify the top content-rating groups.

print("Top content rating by number of apps:")
print(content_counts.head())

print("\nTop content rating by total installs:")
print(content_installs.head())
```

Bar charts were selected because they make it easy to compare the number of applications and total installs across different content-rating groups. They help identify which groups contain the most applications and which groups account for the highest total installs. &#x20;


###### 2. What is/are the insight(s) found from the chart?

The charts help identify the content-rating groups with the highest number of applications and total installs. A group with more applications has a larger representation in the dataset, while a group with higher total installs may have greater overall download activity.  &#x20;

The exact leading groups should be identified from the generated charts and calculated results. The group with the most applications may not necessarily be the group with the highest total installs.


###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can help developers understand the distribution of apps and user demand across content-rating groups. This information can support decisions about target audiences, app features, content suitability, and marketing strategies.   &#x20;

However, focusing on a group with fewer applications does not automatically guarantee business growth. Similarly, high total installs do not necessarily indicate higher revenue or user satisfaction. Developers should consider market demand, competition, ratings, and the intended audience before making business decisions.


#####Q).How do free and paid apps differ in terms of ratings, reviews, and installs?


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert columns to numeric
playstore_data['Rating'] = pd.to_numeric(playstore_data['Rating'], errors='coerce')
playstore_data['Reviews'] = pd.to_numeric(playstore_data['Reviews'], errors='coerce')

playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Group data by app type
type_comparison = playstore_data.groupby('Type').agg({
    'Rating': 'mean',
    'Reviews': 'mean',
    'Installs': 'mean'
}).reset_index()

print(type_comparison)
```

```python
# 1. Compare average ratings

plt.figure(figsize=(8, 5))
sns.barplot(data=type_comparison, x='Type', y='Rating')

plt.title('Average Ratings: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.show()
```

```python
# 2. Compare average number of reviews

plt.figure(figsize=(8, 5))
sns.barplot(data=type_comparison, x='Type', y='Reviews')

plt.title('Average Reviews: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Number of Reviews')
plt.show()
```

```python
# 3. Compare average number of installs

plt.figure(figsize=(8, 5))
sns.barplot(data=type_comparison, x='Type', y='Installs')

plt.title('Average Installs: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Number of Installs')
plt.show()
```

###### 1. Why did you pick the specific chart?

Bar charts were selected because they make it easy to compare free and paid applications across three important measures: average ratings, average number of reviews, and average number of installs. These charts help identify differences in user satisfaction, engagement, and popularity between the two app types.

###### 2. What is/are the insight(s) found from the chart?




The charts help compare free and paid apps in terms of their average ratings, reviews, and installs. Higher average ratings may indicate better user satisfaction, while higher average review counts and installs may indicate greater user engagement and popularity.

The exact findings should be based on the calculated values and generated charts. We should not assume that free apps perform better than paid apps, or vice versa, without examining the results.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can help developers make informed decisions about app pricing, product development, and marketing strategies. If free apps show higher install counts, developers may consider using a free model to attract more users and explore suitable monetization options. If paid apps achieve higher ratings, developers can investigate which features or services users value enough to pay for.

However, paid apps may face challenges in attracting users because of the upfront cost, while free apps may face difficulties generating revenue without effective monetization. These are potential challenges, not conclusions that can be drawn from the charts alone.

Businesses should balance pricing, app quality, user satisfaction, and revenue potential when choosing a pricing strategy.
