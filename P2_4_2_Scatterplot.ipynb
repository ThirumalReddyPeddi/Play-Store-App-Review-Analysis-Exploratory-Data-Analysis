##### Q).What is the relationship between app ratings and the number of reviews?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select required columns
data = playstore_data[['Rating', 'Reviews']].copy()

# Convert columns into numeric format
data['Rating'] = pd.to_numeric(
    data['Rating'], errors='coerce'
)

data['Reviews'] = pd.to_numeric(
    data['Reviews'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

# Remove missing and invalid values
data = data.dropna()

data = data[
    data['Rating'].between(1, 5) &
    (data['Reviews'] > 0)
]

# Create scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=data,
    x='Reviews',
    y='Rating',
    alpha=0.5
)

plt.xscale('log')
plt.title('Relationship Between App Ratings and Number of Reviews')
plt.xlabel('Number of Reviews (Log Scale)')
plt.ylabel('App Rating')
plt.ylim(1, 5)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Calculate correlation
correlation = data['Rating'].corr(data['Reviews'])

print("Correlation:", round(correlation, 3))
```

###### 1. Why did you pick the specific chart?

A scatter plot was selected because it helps visualize the relationship between app ratings and the number of reviews. Each point represents an individual app, making it easier to identify patterns, clusters, and outliers. It also helps determine whether apps with a higher number of reviews tend to receive higher or lower ratings.

###### 2. What is/are the insight(s) found from the chart?

The scatter plot illustrates the relationship between app ratings and the number of reviews. The correlation coefficient helps determine the direction and strength of this relationship. The chart also helps identify apps with a high number of reviews but relatively low ratings, which may indicate potential user dissatisfaction.

Overall, the relationship should be interpreted based on the observed scatter pattern and calculated correlation value. A higher number of reviews does not necessarily indicate a higher app rating.
