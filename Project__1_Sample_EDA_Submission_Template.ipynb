
#####Q).How do free and paid apps differ in terms of ratings, reviews, and installs?


````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert columns to numeric
playstore_data['Rating'] = pd.to_numeric(playstore_data['Rating'], errors='coerce')
playstore_data['Reviews'] = pd.to_numeric(playstore_data['Reviews'], errors='coerce')

playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Group data by app type
type_comparison = playstore_data.groupby('Type').agg({
    'Rating': 'mean',
    'Reviews': 'mean',
    'Installs': 'mean'
}).reset_index()

print(type_comparison)
````

````python
# 1. Compare average ratings

plt.figure(figsize=(8, 5))
sns.barplot(data=type_comparison, x='Type', y='Rating')

plt.title('Average Ratings: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.show()
````

````python
# 2. Compare average number of reviews

plt.figure(figsize=(8, 5))
sns.barplot(data=type_comparison, x='Type', y='Reviews')

plt.title('Average Reviews: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Number of Reviews')
plt.show()
````

````python
# 3. Compare average number of installs

plt.figure(figsize=(8, 5))
sns.barplot(data=type_comparison, x='Type', y='Installs')

plt.title('Average Installs: Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Average Number of Installs')
plt.show()
````

###### 1. Why did you pick the specific chart?

Bar charts were selected because they make it easy to compare free and paid applications across three important measures: average ratings, average number of reviews, and average number of installs. These charts help identify differences in user satisfaction, engagement, and popularity between the two app types.

###### 2. What is/are the insight(s) found from the chart?




The charts help compare free and paid apps in terms of their average ratings, reviews, and installs. Higher average ratings may indicate better user satisfaction, while higher average review counts and installs may indicate greater user engagement and popularity.

The exact findings should be based on the calculated values and generated charts. We should not assume that free apps perform better than paid apps, or vice versa, without examining the results.

Yes, these insights can help developers make informed decisions about app pricing, product development, and marketing strategies. If free apps show higher install counts, developers may consider using a free model to attract more users and explore suitable monetization options. If paid apps achieve higher ratings, developers can investigate which features or services users value enough to pay for.

#####Q).How does the size of an application vary across different categories?

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert app size to MB
def convert_size(size):
    size = str(size).strip()

    if size in ['nan', 'Varies with device']:
        return None

    try:
        if size.endswith('M'):
            return float(size[:-1])
        elif size.endswith('k'):
            return float(size[:-1]) / 1024
    except ValueError:
        return None

    return None

playstore_data['Size_MB'] = playstore_data['Size'].apply(convert_size)

# Remove rows with missing category or size
size_df = playstore_data.dropna(subset=['Category', 'Size_MB'])

# Order categories by median app size
category_order = (
    size_df.groupby('Category')['Size_MB']
    .median()
    .sort_values(ascending=False)
    .index
)

# Create box plot
plt.figure(figsize=(18, 9))
sns.boxplot(
    data=size_df,
    x='Category',
    y='Size_MB',
    order=category_order
)

plt.title('App Size Distribution Across Categories')
plt.xlabel('App Category')
plt.ylabel('App Size (MB)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

###### 1. Why did you pick the specific chart?

A box plot was selected because it shows the distribution of application sizes across different categories. It helps compare the median size, variation in sizes, and potential outliers, making it easier to identify categories with typically larger or smaller applications.  &#x20;


###### 2. What is/are the insight(s) found from the chart?

The chart helps identify differences in app sizes across categories. Categories with higher median sizes tend to contain larger applications, while categories with lower median sizes tend to contain smaller applications. The spread of each box indicates how much app sizes vary within a category, and points beyond the whiskers may represent unusually large or small apps.  &#x20;

The specific categories with the largest and smallest typical app sizes should be identified from the generated chart.


###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can help developers optimize application size and improve user experience. Very large apps may discourage some users from downloading or updating them, while excessively reducing size could remove useful features if optimization is not handled carefully. Developers should balance app size, functionality, and performance based on user needs.  &#x20;

````python
# We can answer this by combining the Play Store dataset (df, containing app categories) with the User Reviews dataset (df_reviews, containing review sentiments).
# We will count positive and negative reviews for each category and compare the results using a grouped bar chart.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Keep the app category from the Play Store dataset
app_categories = playstore_data[['App', 'Category']].drop_duplicates()

# Combine app categories with user reviews
reviews_with_category = user_reviews_data.merge(
    app_categories,
    on='App',
    how='inner'
)

# Keep positive and negative reviews
feedback = reviews_with_category[
    reviews_with_category['Sentiment'].isin(['Positive', 'Negative'])
]

# Count positive and negative reviews by category
category_sentiment = (
    feedback.groupby(['Category', 'Sentiment'])
    .size()
    .reset_index(name='Review_Count')
)

# Select the top 10 categories by total positive and negative reviews
top_categories = (
    category_sentiment.groupby('Category')['Review_Count']
    .sum()
    .nlargest(10)
    .index
)

plot_data = category_sentiment[
    category_sentiment['Category'].isin(top_categories)
]

# Create grouped bar chart
plt.figure(figsize=(15, 7))
sns.barplot(
    data=plot_data,
    x='Category',
    y='Review_Count',
    hue='Sentiment'
)

plt.title('Positive and Negative User Reviews by App Category')
plt.xlabel('App Category')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

#####Q). pair plot helps visualize the relationships between multiple numerical variables in the Google Play Store dataset

````python
# A pair plot helps visualize the relationships between multiple numerical variables in the Google Play Store dataset.
# It displays scatter plots for pairs of variables and distribution plots along the diagonal.


import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Select numerical columns
pair_df = playstore_data[['Rating', 'Reviews', 'Installs', 'Price']].copy()

# Clean the data
pair_df['Rating'] = pd.to_numeric(pair_df['Rating'], errors='coerce')

pair_df['Reviews'] = pd.to_numeric(
    pair_df['Reviews'], errors='coerce'
)

pair_df['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
      .str.replace(',', '', regex=False)
      .str.replace('+', '', regex=False),
    errors='coerce'
)

pair_df['Price'] = pd.to_numeric(
    playstore_data['Price'].astype(str)
      .str.replace('$', '', regex=False),
    errors='coerce'
)

# Keep valid ratings and remove missing values
pair_df = pair_df[
    pair_df['Rating'].between(1, 5)
].dropna()

# Use a sample to make the plot easier to render
pair_df = pair_df.sample(
    n=min(1000, len(pair_df)),
    random_state=42
)

# Create pair plot
sns.pairplot(pair_df, diag_kind='hist')
plt.show()
````

A pair plot was selected because it allows us to visualize relationships between multiple numerical variables simultaneously. It helps compare ratings, reviews, installs, and prices using scatter plots, while the diagonal histograms show the distribution of each variable.
