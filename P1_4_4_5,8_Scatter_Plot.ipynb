```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Convert columns to numeric
playstore_data['Reviews'] = pd.to_numeric(playstore_data['Reviews'], errors='coerce')

playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Remove rows with missing values
plot_df = playstore_data.dropna(subset=['Reviews', 'Installs'])

# Create scatter plot
plt.figure(figsize=(12, 7))
sns.scatterplot(data=plot_df, x='Reviews', y='Installs', alpha=0.5)

plt.title('Relationship Between Number of Reviews and Installs')
plt.xlabel('Number of Reviews')
plt.ylabel('Number of Installs')
plt.tight_layout()
plt.show()
```

```python
correlation = plot_df['Reviews'].corr(plot_df['Installs'])
print("Correlation coefficient:", correlation)
```

#####Q).Is there a relationship between app ratings and the number of reviews?

```python
# Chart - 5 visualization code
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert columns to numeric
playstore_data['Rating'] = pd.to_numeric(playstore_data['Rating'], errors='coerce')
playstore_data['Reviews'] = pd.to_numeric(playstore_data['Reviews'], errors='coerce')

# Remove missing values
plot_df = playstore_data.dropna(subset=['Rating', 'Reviews'])

# Keep valid ratings between 1 and 5
plot_df = plot_df[plot_df['Rating'].between(1, 5)]

# Create scatter plot
plt.figure(figsize=(12, 7))
sns.scatterplot(
    data=plot_df,
    x='Reviews',
    y='Rating',
    alpha=0.5
)

plt.title('Relationship Between App Ratings and Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('App Rating')
plt.tight_layout()
plt.show()
```

```python
correlation = plot_df['Rating'].corr(plot_df['Reviews'])
print("Correlation coefficient:", correlation)
```

###### 1. Why did you pick the specific chart?

A scatter plot was selected because it helps visualize the relationship between app ratings and the number of reviews. Each point represents an app, making it easier to observe patterns and determine whether apps with more reviews tend to receive higher or lower ratings.

###### 2. What is/are the insight(s) found from the chart?

The scatter plot helps us understand how app ratings vary with the number of reviews. If the points show a clear upward or downward trend, it may indicate a positive or negative relationship between the variables. If the points are widely scattered, the relationship may be weak.

The actual relationship should be confirmed using the correlation coefficient calculated from the dataset.

###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can help businesses understand user satisfaction and engagement. Apps with many reviews can provide valuable feedback for identifying strengths and areas for improvement. Developers can use this information to improve app features, performance, and user experience.

If apps receive many reviews but have low ratings, it may indicate user dissatisfaction that needs attention. Ignoring recurring complaints may lead to reduced user trust and lower long-term retention. However, a large number of reviews does not automatically mean an app has a high or low rating. &#x20;

Businesses should analyze ratings along with review content and sentiment to make informed decisions and improve user satisfaction.


#### Chart - 8

##### Q).Does app price have any relationship with app ratings or popularity?



```python
# Chart - 8 visualization code
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert columns to numeric
playstore_data['Price'] = pd.to_numeric(
    playstore_data['Price'].astype(str).str.replace('$', '', regex=False),
    errors='coerce'
)

playstore_data['Rating'] = pd.to_numeric(playstore_data['Rating'], errors='coerce')
playstore_data['Reviews'] = pd.to_numeric(playstore_data['Reviews'], errors='coerce')

playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Keep rows with valid price, rating, reviews, and installs
price_df = playstore_data.dropna(
    subset=['Price', 'Rating', 'Reviews', 'Installs']
)

# Remove invalid negative prices
price_df = price_df[price_df['Price'] >= 0]

# Price vs Rating
plt.figure(figsize=(10, 6))
sns.scatterplot(data=price_df, x='Price', y='Rating', alpha=0.5)

plt.title('App Price vs Rating')
plt.xlabel('App Price ($)')
plt.ylabel('App Rating')
plt.tight_layout()
plt.show()
```

```python
# Compare price with popularity (installs)

plt.figure(figsize=(10, 6))
sns.scatterplot(data=price_df, x='Price', y='Installs', alpha=0.5)

plt.yscale('log')
plt.title('App Price vs Number of Installs')
plt.xlabel('App Price ($)')
plt.ylabel('Number of Installs (log scale)')
plt.tight_layout()
plt.show()
```

```python
# Calculate correlations

print("Price vs Rating:",
      price_df['Price'].corr(price_df['Rating']))

print("Price vs Reviews:",
      price_df['Price'].corr(price_df['Reviews']))

print("Price vs Installs:",
      price_df['Price'].corr(price_df['Installs']))
```

###### 1. Why did you pick the specific chart?

Scatter plots were selected because they help visualize the relationship between app prices and ratings or installs. They make it easier to identify patterns and observe whether higher-priced apps tend to receive higher ratings or attract more downloads.   &#x20;


###### 2. What is/are the insight(s) found from the chart?

The charts help us examine whether app prices are associated with user ratings and popularity. If higher-priced apps have fewer installs, this may suggest that the upfront cost could be a barrier for some users. If ratings remain similar across different price levels, price alone may not explain differences in user satisfaction.  &#x20;

The actual conclusions should be based on the generated charts and calculated correlation values rather than assumptions. &#x20;


###### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes, these insights can help developers make informed pricing decisions. Understanding the relationship between price, ratings, and installs can help businesses choose suitable prices, improve app value, and develop effective monetization strategies.  &#x20;

If higher prices are associated with fewer installs, developers may need to reconsider their pricing or demonstrate the additional value their apps provide. However, lowering prices does not automatically increase revenue, and higher prices do not necessarily result in higher ratings.

Businesses should balance pricing, app quality, user satisfaction, competition, and revenue potential when deciding on a pricing strategy.
