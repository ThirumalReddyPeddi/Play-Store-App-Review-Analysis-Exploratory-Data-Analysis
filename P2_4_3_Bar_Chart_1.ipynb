#### Chart - 3 : **Bar Chart**

##### Q).Which apps or categories have the highest total installs?

```python
# Chart - 3 visualization code

# Useful for comparing values across categories.
```

```python
# Convert Installs into numeric format
playstore_data['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)

# Get top 10 apps based on installs
top_apps = (
    playstore_data.groupby('App')['Installs']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# Create bar plot
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_apps.index,
    y=top_apps.values
)

plt.title('Top 10 Apps Based on Total Installs')
plt.xlabel('App Name')
plt.ylabel('Total Installs')
plt.xticks(rotation=75, ha='right')
plt.tight_layout()
plt.show()

print(top_apps)
```

```python
# Calculate total installs for each category
category_installs = (
    playstore_data.groupby('Category')['Installs']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# Create bar plot
plt.figure(figsize=(12, 6))

sns.barplot(
    x=category_installs.index,
    y=category_installs.values
)

plt.title('Top 10 Categories Based on Total Installs')
plt.xlabel('App Category')
plt.ylabel('Total Installs')
plt.xticks(rotation=75, ha='right')
plt.tight_layout()
plt.show()

print(category_installs)
```

###### 1. Why did you pick the specific chart?

A bar plot was selected because it provides a clear visual comparison of total app installations across different apps or categories. The height of each bar represents the total number of installs, making it easy to identify the most and least popular apps or categories. It helps us understand the distribution of app popularity and compare installation levels effectively.

###### 2. What is/are the insight(s) found from the chart?

The bar plot helps identify which apps and categories have the highest total installations. It highlights the most popular apps and categories that attract a larger number of users. Categories with higher total installs may indicate greater user demand and market interest. It also helps identify categories with relatively fewer installations, which may have opportunities for improvement or growth.
