#####Q). pair plot helps visualize the relationships between multiple numerical variables in the Google Play Store dataset

```python
# A pair plot helps visualize the relationships between multiple numerical variables in the Google Play Store dataset.
# It displays scatter plots for pairs of variables and distribution plots along the diagonal.


import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Select numerical columns
pair_df = playstore_data[['Rating', 'Reviews', 'Installs', 'Price']].copy()

# Clean the data
pair_df['Rating'] = pd.to_numeric(pair_df['Rating'], errors='coerce')

pair_df['Reviews'] = pd.to_numeric(
    pair_df['Reviews'], errors='coerce'
)

pair_df['Installs'] = pd.to_numeric(
    playstore_data['Installs'].astype(str)
      .str.replace(',', '', regex=False)
      .str.replace('+', '', regex=False),
    errors='coerce'
)

pair_df['Price'] = pd.to_numeric(
    playstore_data['Price'].astype(str)
      .str.replace('$', '', regex=False),
    errors='coerce'
)

# Keep valid ratings and remove missing values
pair_df = pair_df[
    pair_df['Rating'].between(1, 5)
].dropna()

# Use a sample to make the plot easier to render
pair_df = pair_df.sample(
    n=min(1000, len(pair_df)),
    random_state=42
)

# Create pair plot
sns.pairplot(pair_df, diag_kind='hist')
plt.show()
```

###### 1. Why did you pick the specific chart?

A pair plot was selected because it allows us to visualize relationships between multiple numerical variables simultaneously. It helps compare ratings, reviews, installs, and prices using scatter plots, while the diagonal histograms show the distribution of each variable.

###### 2. What is/are the insight(s) found from the chart?

The pair plot helps identify possible relationships, clusters, and outliers among app ratings, reviews, installs, and prices. It can show whether apps with more reviews tend to have more installs and whether ratings or prices appear related to popularity. The actual insights should be based on the patterns visible in the generated chart; the plot alone does not prove causation.
