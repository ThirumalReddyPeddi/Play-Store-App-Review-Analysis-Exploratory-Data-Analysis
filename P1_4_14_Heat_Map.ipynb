#####Q). Provide a Correlation Heatmap shows the relationships between numerical variables in the Google Play Store dataset

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Select numerical columns
corr_df = playstore_data[['Rating', 'Reviews', 'Installs', 'Price']].copy()

# Ensure values are numeric
for col in corr_df.columns:
    corr_df[col] = pd.to_numeric(corr_df[col], errors='coerce')

# Calculate correlations
correlation_matrix = corr_df.corr()

# Plot heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap of Play Store App Features')
plt.tight_layout()
plt.show()
```

###### 1. Why did you pick the specific chart?

A correlation heatmap was selected because it visually represents the strength and direction of relationships between numerical variables. The color scale and correlation values make it easier to identify positive, negative, and weak linear relationships among app ratings, reviews, installs, and prices. &#x20;


###### 2. What is/are the insight(s) found from the chart?

The heatmap helps identify which numerical variables are positively or negatively correlated and which have weak linear relationships. These patterns can help us understand how app ratings, reviews, installs, and prices are associated with one another.  &#x20;

The specific insights should be based on the correlation values displayed in the generated heatmap. Correlation does not prove that one variable causes changes in another.
